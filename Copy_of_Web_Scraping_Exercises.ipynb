**TASK: Use requests library and BeautifulSoup to connect to http://quotes.toscrape.com/ and get the HMTL text from the homepage.**

```python
url='https://quotes.toscrape.com/'
response=requests.get(url)
soup=BeautifulSoup(response.text,'html.parser')
```

**TASK: Get the names of all the authors on the first page.**

```python
auth=soup.find_all('small',class_='author')
for i in auth:
  print(i.text)
```

**TASK: Create a list of all the quotes on the first page.**

```python
qoute=[]
qoutes=soup.find_all('span',class_='text')
for i in qoutes:
  qoute.append(i.text)
print(qoute)
```

**TASK: Notice how there is more than one page, and subsequent pages look like this http://quotes.toscrape.com/page/2/. Use what you know about for loops and string concatenation to loop through all the pages and get all the unique authors on the website.

```python
import requests
from bs4 import BeautifulSoup

authors = set()  # stores unique authors

for page in range(1, 11):

    url = f"http://quotes.toscrape.com/page/{page}/"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    author_tags = soup.select(".author")

    for author in author_tags:
        authors.add(author.text)

print(authors)
print(f"\nTotal Unique Authors: {len(authors)}")
```
